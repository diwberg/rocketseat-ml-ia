# Pré-processamento dos datasets

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

## Classificação - Airline

```python
train = pd.read_csv('dados/train.csv')
test = pd.read_csv('dados/test.csv')

# Amostra do treino para viabilizar o desafio
train = train.sample(10000, random_state=42)
```

```python
def preparar_airline(df):
    df = df.drop(columns=['Unnamed: 0', 'id'])
    df['Arrival Delay in Minutes'] = df['Arrival Delay in Minutes'].fillna(df['Departure Delay in Minutes'])
    df['satisfaction'] = (df['satisfaction'] == 'satisfied').astype(int)
    df = pd.get_dummies(df, columns=['Gender', 'Customer Type', 'Type of Travel', 'Class'], drop_first=True, dtype=int)
    return df

train = preparar_airline(train)
test = preparar_airline(test)
test = test[train.columns]
```

```python
features = train.columns.drop('satisfaction')
scaler = StandardScaler()
train[features] = scaler.fit_transform(train[features])
test[features] = scaler.transform(test[features])
```

```python
train.to_csv('dados/airline_train_clean.csv', index=False)
test.to_csv('dados/airline_test_clean.csv', index=False)
train.shape, test.shape
```

## Regressão - Taxi

```python
taxi = pd.read_csv('dados/taxi_trip_pricing.csv')

# Remove linhas sem a variável alvo
taxi = taxi.dropna(subset=['Trip_Price'])

numericas = taxi.select_dtypes('number').columns.drop('Trip_Price')
categoricas = ['Time_of_Day', 'Day_of_Week', 'Traffic_Conditions', 'Weather']

taxi[numericas] = taxi[numericas].fillna(taxi[numericas].median())
taxi[categoricas] = taxi[categoricas].fillna(taxi[categoricas].mode().iloc[0])
taxi = pd.get_dummies(taxi, columns=categoricas, drop_first=True, dtype=int)
```

```python
taxi_train, taxi_test = train_test_split(taxi, test_size=0.2, random_state=42)

features = taxi.columns.drop('Trip_Price')
scaler = StandardScaler()
taxi_train[features] = scaler.fit_transform(taxi_train[features])
taxi_test[features] = scaler.transform(taxi_test[features])
```

```python
taxi_train.to_csv('dados/taxi_train_clean.csv', index=False)
taxi_test.to_csv('dados/taxi_test_clean.csv', index=False)
taxi_train.shape, taxi_test.shape
```
