```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Carregar os dados
df = pd.read_csv('housing.csv')
df.head()
```

```python
df.info()
```

```python
# Pré-processamento
binarias = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea']
df[binarias] = df[binarias].replace({'yes': 1, 'no': 0})
df = pd.get_dummies(df, columns=['furnishingstatus'], dtype=int)

X = StandardScaler().fit_transform(df)
```

```python
# PCA mantendo 90% da variância
pca = PCA(n_components=0.9)
X_pca = pca.fit_transform(X)
print('Componentes:', pca.n_components_)
print('Variância explicada:', pca.explained_variance_ratio_)
print('Variância acumulada:', pca.explained_variance_ratio_.sum())
```

```python
# Visualização nos dois primeiros componentes
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=df['price'])
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.colorbar(label='price')
plt.show()
```

```python
# Erro de reconstrução
X_reconstruido = pca.inverse_transform(X_pca)
erro = np.mean((X - X_reconstruido) ** 2)
print('Erro de reconstrução (MSE):', erro)
```

```python
# Principais fatores
cargas = pd.DataFrame(pca.components_[:2].T, index=df.columns, columns=['PC1', 'PC2'])
cargas.sort_values('PC1', ascending=False)
```

## Interpretação

Para manter 90% da variância, o PCA precisou de 11 componentes, contra 15 colunas originais. A redução foi pequena porque a variância está bem distribuída: o primeiro componente explica 22% e os outros ficam entre 4% e 11% cada. O erro de reconstrução (MSE) com esses 11 componentes foi de cerca de 0,08 nos dados padronizados.

O PC1 representa o "tamanho e padrão" do imóvel. Ele tem peso alto em preço, área, banheiros, ar-condicionado, quartos, vagas e andares, e peso negativo em casa sem mobília. Ou seja, quanto maior o PC1, maior e mais cara a casa. No gráfico isso aparece como as cores de preço mais alto concentradas à direita.

O PC2 representa principalmente o status de mobília, separando as casas semimobiliadas das mobiliadas e das sem mobília.
