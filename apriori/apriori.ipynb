```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
# Carregar e explorar os dados
df = pd.read_csv('game_sales.csv')
df.head()
```

```python
df.info()
```

```python
# Quantidade de jogos únicos
df['Game Name'].nunique()
```

```python
# Jogos mais vendidos
df['Game Name'].value_counts().head(10)
```

```python
# Porcentagem de ocorrência de cada jogo em relação aos clientes
(df.groupby('Game Name')['Client ID'].nunique() / df['Client ID'].nunique() * 100).sort_values(ascending=False)
```

```python
# Preparar os dados (uma linha por cliente, uma coluna por jogo)
cesta = pd.crosstab(df['Client ID'], df['Game Name']) > 0
cesta.head()
```

```python
# Apriori
itens_frequentes = apriori(cesta, min_support=0.35, use_colnames=True)
regras = association_rules(itens_frequentes, metric='lift', min_threshold=1)
```

```python
# Regras ordenadas por lift
regras[['antecedents', 'consequents', 'support', 'confidence', 'lift']].sort_values('lift', ascending=False).head(10)
```

## Insights

A base tem 33 jogos e 591 clientes. As vendas são bem distribuídas: o mais vendido (Call of Duty: Modern Warfare) aparece para 56% dos clientes e o menos vendido para 48%. Cada cliente comprou em média 17 jogos, então os jogos aparecem juntos com muita frequência. Por isso foi preciso um suporte mínimo alto (35%) para chegar a um número de regras que dá para analisar.

As regras mais fortes, pelo lift, foram:

- Far Cry 6 e Roblox (lift 1,39, confiança de 73% de Far Cry 6 para Roblox)
- Fortnite e Resident Evil 4 (lift 1,38)
- Red Dead Redemption 2 e Fall Guys (lift 1,36)
- Resident Evil 4 e Fall Guys (lift 1,36)
- Dota 2 e Roblox (lift 1,36)

Todas têm lift acima de 1, ou seja, esses pares saem juntos mais do que o esperado por acaso, mas o efeito é moderado.

- Bundles: Fortnite + Resident Evil 4 + Fall Guys forma um trio que aparece ligado nas regras e pode virar um pacote.
