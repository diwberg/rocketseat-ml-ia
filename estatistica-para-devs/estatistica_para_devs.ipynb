```python
import pandas as pd
import matplotlib.pyplot as plt

dict_faturamento = {
    'data_ref': [
        '2023-01-01',
        '2020-02-01',
        '2021-03-01',
        '2022-04-01',
        '2023-05-01',
        '2023-06-01',
        '2020-07-01',
        '2021-08-01',
        '2022-09-01',
        '2023-10-01',
        '2022-11-01',
        '2023-12-01',
    ],
    'valor': [
        400000,
        890000,
        760000,
        430000,
        920000,
        340000,
        800000,
        500000,
        200000,
        900000,
        570000,
        995000,
    ]
}
```

```python
df = pd.DataFrame(dict_faturamento)
df['data_ref'] = pd.to_datetime(df['data_ref'])
df = df.sort_values('data_ref')
```

```python
# Média das vendas
print(df['valor'].mean())
```

```python
# Gráfico de barras vertical
plt.bar(df['data_ref'].dt.strftime('%m/%Y'), df['valor'])
plt.xticks(rotation=45)
plt.show()
```

```python
# Gráfico de linhas
plt.plot(df['data_ref'], df['valor'])
plt.show()
```
