```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
```

```python
df = pd.read_csv('synthetic_cell_data.csv')
X = df.drop(columns=['cell_type'])
cores = df['cell_type'].astype('category').cat.codes
```

```python
# t-SNE de 50 para 2 dimensões
X_2d = TSNE(n_components=2, random_state=42).fit_transform(X)
```

```python
# Gráfico 2D
plt.scatter(X_2d[:, 0], X_2d[:, 1], c=cores)
plt.xlabel('Componente 1')
plt.ylabel('Componente 2')
plt.show()
```

```python
# t-SNE com 3 componentes para o gráfico 3D
X_3d = TSNE(n_components=3, random_state=42).fit_transform(X)
```

```python
# Gráfico 3D
ax = plt.figure().add_subplot(projection='3d')
ax.scatter(X_3d[:, 0], X_3d[:, 1], X_3d[:, 2], c=cores)
ax.set_xlabel('Componente 1')
ax.set_ylabel('Componente 2')
ax.set_zlabel('Componente 3')
plt.show()
```

## Análise visual

No gráfico 2D, os 5 tipos celulares formam 5 agrupamentos claros, cada um com uma cor predominante. Há alguns pontos de outras cores dentro de cada grupo e uma leve sobreposição nas bordas onde os grupos se encostam, mas a separação é bem visível.

No gráfico 3D os grupos também aparecem separados por cor, mas a leitura é mais difícil por causa da perspectiva, que deixa alguns grupos uns na frente dos outros.

No geral, a expressão gênica separa bem os tipos celulares, e o t-SNE conseguiu mostrar isso em poucas dimensões.
